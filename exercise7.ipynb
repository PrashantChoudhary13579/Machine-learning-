Read the Colleges dataset and clean all the numeric attributes.
From all marks based attributes, remove % sign from different attributes. In SAT/ACT column, Keep higher value.
From all cost based attributes, Remove $ sign and commas.
Remove any null value in any attribute.

```python
import pandas as pd
```

```python
path ="Top 100 Private Colleges.2003.csv"
colleges = pd.read_csv(path)

colleges_copy = colleges.copy()

colleges.head()
```

```python
# remove % sign
def clean_percentage(value):
    if pd.isnull(value):
        return value
    return str(value).replace('%', '')
```

```python
# remove $ and commas
def clean_cost(value):
    if pd.isnull(value):
        return value
    return str(value).replace('$', '').replace(',', '')
```

```python
# keep higher value in SAT/ACT column
def clean_sat_act(value):
    if pd.isnull(value):
        return value

    parts = str(value).replace('/', ' ').split()
    numbers = []

    for p in parts:
        cleaned = ''.join(ch for ch in p if ch.isdigit())
        if cleaned != '':
            numbers.append(int(cleaned))

    if numbers:
        return max(numbers)

    return value
```

```python
# cleaning
for col in colleges_copy.columns:
    
    # Percentage columns
    if colleges_copy[col].astype(str).str.contains('%', regex=False).any():
        print("%", col)
        colleges_copy[col] = colleges_copy[col].apply(clean_percentage)
    
    # Cost columns
    if colleges_copy[col].astype(str).str.contains('$', regex=False).any():
        print("$", col)
        colleges_copy[col] = colleges_copy[col].apply(clean_cost)
    
    # SAT/ACT columns
    if "SAT" in col or "ACT" in col:
        print("SAT", col)
        colleges_copy[col] = colleges_copy[col].apply(clean_sat_act)
```

```python
# convert string values to numeric after removal of dollar and so
for col in colleges_copy.columns:
    colleges_copy[col] = pd.to_numeric(colleges_copy[col], errors='ignore')

colleges_copy.isna().sum()
```

```python
# null values
colleges_copy = colleges_copy.dropna(axis=1, how="any")
print(len(colleges_copy))
```
