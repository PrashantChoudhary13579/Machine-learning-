### 1. Find correlations between various features and grade_mean and see which features have a significant impact on it.

```python
df['Grade_mean'] = df[['G1','G2','G3']].mean(axis = 1)
print(df['Grade_mean'])
```

```python
grade_corr = df.corr(numeric_only = True)['GradeMean'].drop('GradeMean')
print(grade_corr.sort_values(ascending=False))
```

### 2. Find out who got better grades (Boys or Girls). Similarly Find out who got better grades age wise.

```python
avg_gender = df.groupby('sex')['G3'].mean()
print(avg_gender)
```

```python
import matplotlib.pyplot as plt

avg_gender.plot(kind='bar')
plt.title("Average Grades by Gender")
plt.ylabel("Average Grade")
plt.show()
```

```python
avg_age = df.groupby('age')['G3'].mean()
print(avg_age)
```

```python
avg_age.plot(kind='bar')
plt.title("Average Grades by Age")
plt.xlabel("Age")
plt.ylabel("Average Grade")
plt.show()
```
