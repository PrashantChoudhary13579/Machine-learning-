```python
# finding all directors 

directors = set()
for d in data:
    if d["director"]:
        directors.add(d["director"])

print(len(directors))
```

```python
# counts of releases acc. to country
country_counts = {}
country = {}
for d in data:
    country = d["country"]

    if country:
        country_counts[country] = country_counts.get(country,0) +1 

print(country_counts)
sorted_data = sorted(
    country_counts.items(), key= lambda x:x[1], reverse= True
)
top20 = sorted_data[:20]
print(top20)

countries = [x[0] for x in top20 ]
counts = [x[1] for x in top20 ]
```

```python

import matplotlib.pyplot as plt

# countries = list(country_counts.keys())[:10]
# counts =    list(country_counts.values())[:10]

plt.figure(figsize=(12,6))
plt.bar(countries, counts)
plt.xticks(rotation = 90)
plt.xlabel("Countries")
plt.ylabel("No. of movies/shows")
plt.title ("Overall analysis")
```

```python
import csv

with open("Movies.csv" ,"r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    data = list(reader)

for i in range(10):
    print(data[i])
```

```python
tv_ma = [ d for d in data if d["rating"]=="TV-MA"]
print(len(tv_ma))

count_tv = {}

for d in data:
    t = d["rating"]
    if t not in count_tv:
        count_tv[t] = 0
    count_tv[t] += 1

print(count_tv)

```

```python
# ploting into a histogram or bar graph
import matplotlib.pyplot as plt
name = count_tv.keys()
value = count_tv.values()

plt.figure(figsize=(12,4))
plt.bar(name, value)
plt.xticks(rotation = 50)
plt.xlabel("Rating type")
plt.ylabel("No. of movies rated ")
plt.show()
```

```python
# find the longest duration of a movie. 
print(data[1]["duration"])
print(type(data[1]["duration"]))

max_time = 4000
longest_movie = None

for d in data:
    if d["type"] =="Movie":
        duration = int(d["duration"].split()[0])
        if max_time > duration :
            max_time = duration
            longest_movie = d["title"]
        


print(f"{longest_movie} is the smallest movie of {max_time} mins ")
```
