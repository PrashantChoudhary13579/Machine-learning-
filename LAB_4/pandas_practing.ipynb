```python
import numpy as np
import pandas as pd

array = np.ones(5, dtype=([('first', int), ('second','U4'), ('third',float)]))
print(array)
print(array.shape)
print(array['first'])
print(array['second'])
print(array['third'])
print(array['first']+ array['third'])
```

```python
b = array.view(np.recarray)
print(b)
print(b.first)
print(b.second[3])
print(type(b.third))
print(type(b.third[2]))
print(b.third) # there is a difference between them here 
print(array['third'])   
```

```python
data_val = [[1,2,3],[4,5,6],[7,8,9]]
data_df = pd.DataFrame(data = data_val, index = range(100,103), columns=['A','B','D'])
print(data_df)
# same case , row = range value/array of size data_val.shape[0] , col = columns value 
```

Accessing elements of dataframe

```python
print(data_df.iloc[0,2]) # gives the value , works like matrix indexing
print(data_df.iat[0,2]) # gives the value , works like matrix indexing


print(data_df.loc[100:101,'A':'D']) # need to take the row value and the column value such that at 100 th row and A col value = ?
print(data_df.loc[102,'D'])
print(data_df.at[102,'D'])

# at and iat are faster because they only give one single value
```

```python
data_df['C'] = data_df['A']
data_df['E'] = data_df.index
print(data_df)
print(data_df.shape)
```

Combining data frame

```python
Eno=[100, 101,102, 103, 104,105]
Empname= ['Raja', 'Babu', 'Kumar','Karthik','Rajesh','Raju']

Eno_series = pd.Series(Eno)
Emp_series = pd.Series(Empname)

df = {'Eno' :Eno_series,'Empname':Emp_series}
employee1 = pd.DataFrame(df)

(employee1)
```

```python
Eno1=[106, 107,108, 109, 110]
Empname1= ['Jack', 'John', 'Peter','David','Davis']
Eno_Series1 = pd.Series(Eno1)
Empname_Series1 = pd.Series(Empname1)
df = { 'Eno': Eno_Series1, 'Empname': Empname_Series1 }
employee2 = pd.DataFrame(df)
employee2
```

```python
df_concat = pd.concat([employee1,employee2], ignore_index=True)
df_concat
```

```python
coll = [employee1, employee2]

df_con_key = pd.concat(coll, keys=['Section-A','Section-B'])
print(df_con_key)
```
