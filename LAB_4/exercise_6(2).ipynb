### Q3. Sort the records based on rpc attribute (Research Professional practices).

```python
rpc_sorted_records = df.sort_values(by='rpc', ascending=False,ignore_index = True)
# print(rpc_sorted_records)
rpc_sorted_records.head()
```

### Q4. Find the rank of Panjab University as per rpc attribute.

```python
df_sorted = df.sort_values(by='rpc', ascending=False).reset_index(drop=True)
df_sorted["rank"] = df_sorted.index + 1

college_rank = df_sorted[df_sorted["name"] == "Panjab University"]["rank"]
print("Panjab University Rank acc. to rpc -",college_rank.values[0])
```

### Q5. List the colleges whose rpc value is less than Panjab University and rank is higher(rank value is less).

```python
rpc_pu = df_sorted[df_sorted.name == "Panjab University"]["rpc"].values[0]
rank_pu = df_sorted[df_sorted.name =="Panjab University"]["rank"].values[0]
print(rank_pu)
data  = df_sorted.iloc[rank_pu:][df_sorted["rank"] < rank_pu]

print("All the colleges whose rpc value is less than PU and rank is higher: ", len(data))


print(data["name"])
```

```python
df1 = df.sort_values(by = "rpc",ascending=False ).reset_index(drop=True)

df1["Rank"] = df1.index + 1

pu = df1[df1["name"] == "Panjab University"]
pu_rpc = pu["rpc"].values[0]
pu_rank = pu["rank"].values[0]

result = df1[
    (df1["rank"] < pu_rank) & (df1["rpc"] < pu_rpc)
]
print("rpc value of PU: ",pu_rpc)
print("Rank of PU: ",pu_rank)
print(result["name"])
```

```python

# Sort and rank
df_sorted = df.sort_values(by="rpc", ascending=False).reset_index(drop=True)
df_sorted["rank"] = df_sorted.index + 1

# Get Panjab University data
pu = df_sorted[df_sorted["name"] == "Panjab University"]

pu_rpc = pu["rpc"].values[0]
pu_rank = pu["rank"].values[0]

# Filter
result = df_sorted[
    (df_sorted["rpc"] < pu_rpc) &
    (df_sorted["rank"] < pu_rank)
]

print(result)
```

### Q6. Draw the Histogram for colleges based on tlr values by dividing the overall data in 10 bins

```python
import matplotlib.pyplot as plt

plt.hist(df["tlr"], bins=10, edgecolor="black")
plt.title("Histogram for TLR")
plt.xlabel("tlr values")
plt.ylabel("frequency")
plt.grid(False)
plt.show()
```
