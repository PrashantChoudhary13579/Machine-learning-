### Q1. From all marks based attributes, remove % sign from different attributes. In SAT/ACT column, Keep higher value.


```python
def percentage_removal(value):
    if pd.isnull(value):
        return value

    value = str(value).strip()

    if '%' in value:
        return value.replace('%','')

    if '/' in value:  # case like 99/100
        num, den = value.split('/')
        return (float(num)/float(den))*100

    return value


for col in df.columns:
    if df[col].astype(str).str.contains('%').any():
        df[col] = df[col].apply(percentage_removal)

df.head()
```

```python
import re

def clean_sat_act(value):
    if pd.isna(value):
        return None

    value = str(value)

    numbers = re.findall(r'\d+', value)   

    if numbers:
        numbers = [int(n) for n in numbers]
        return max(numbers)

    return None

df['*SAT or ACT'] = df['*SAT or ACT'].apply(clean_sat_act)
df.head(10)
```

### Q2. From all cost based attributes, Remove $ sign and commas.


```python
def clean_dollar(value):
    if pd.isnull(value):
        return value
    
    value = str(value)
    value = value.replace('$','').replace(',','')
    return value

for col in df.columns:
    if df[col].astype(str).str.contains('\$').any():
        df[col] = df[col].apply(clean_dollar)

df.head(10)
```
