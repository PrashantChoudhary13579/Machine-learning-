### Q1. From all marks based attributes, remove % sign from different attributes. In SAT/ACT column, Keep higher value.


```python
# we have to make a list of columns which contains % sign

def percentage_removal(x):
    if pd.isna(x):
        return None
    
    x = str(x).strip()

    # case 1
    if '%' in x:
        return float(x.replace("%", ""))
    
    if '/' in x:
        num, deno = x.split("/")
        return (float(num) / float(deno))* 100

    try:
        return float(x)
    except:
        return None

percentage_sign =   [
    'Admission Rate',
    '*SAT or ACT',
'4-year Grad. Rate',
'6-year Grad. Rate',
'Quality Rank',
'Need Met', 
'Aid From Grants',
'Non-Need-Based Aid+'
]

for col in percentage_sign:
    df[col] = df[col].apply(percentage_removal)
```
