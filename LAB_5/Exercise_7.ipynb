```python
import pandas as pd

# path = "Top 100 Private Colleges.2003.csv"
colleges = pd.read_csv((r"D:\College_work\ML\LAB_5\Top100_PrivateColleges2003.csv"))

colleges_copy = colleges.copy()

colleges_copy.head()
```

```python
def clean_percentage(value):
    if pd.isnull(value):
        return value

    value = str(value).strip()

    if '%' in value:
        return value.replace('%','')

    if '/' in value:  # case like 99/100
        num, den = value.split('/')
        return (float(num)/float(den))*100

    return value
```

```python
def clean_cost(value):
    if pd.isnull(value):
        return value

    value = str(value)
    value = value.replace('$','').replace(',','')

    return value
```

```python
def clean_sat_act(value):
    if pd.isnull(value):
        return value

    value = str(value)

    parts = value.replace('/', ' ').split()

    numbers = []

    for p in parts:
        if p.isdigit():
            numbers.append(int(p))

    if numbers:
        return max(numbers)

    return None
```

```python
for col in colleges_copy.columns:

    # Percentage columns
    if colleges_copy[col].astype(str).str.contains('%').any():
        colleges_copy[col] = colleges_copy[col].apply(clean_percentage)

    # Cost columns
    if colleges_copy[col].astype(str).str.contains('\$').any():
        colleges_copy[col] = colleges_copy[col].apply(clean_cost)
```

```python
colleges_copy['*SAT or ACT'] = colleges_copy['*SAT or ACT'].apply(clean_sat_act)
```

```python
for col in colleges_copy.columns:
    colleges_copy[col] = pd.to_numeric(colleges_copy[col], errors='ignore')
```
