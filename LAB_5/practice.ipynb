```python
a = train.isnull().sum()/len(train)*100
variables = train.columns
print(variables)

variable = []
for i in range(0,12):
    if a[i] <=  30:   #setting the threshold at 20%
        variable.append(variables[i])

print(variable)

# so we have remove the variables having more than 20% missing values 
```

```python
# low variance filter 
# lets first impute the missing values in the Item_Weight col. using the median value

train["Item_Weight"].fillna(train["Item_Weight"].mean())
# train["Item_Weight"].fillna(train["Outlet_Size"].mode()[0])
```

```python
train["Outlet_Size"].fillna(train["Outlet_Size"].mode()[0])
```

```python
train["Item_Fat_Content"].unique()
```

```python
# now replacing these -  'low fat' , 'LF', 'reg' 

train["Item_Fat_Content"] = train["Item_Fat_Content"].replace({
    "LF":"Low Fat", 
    "reg" :"Regular",
    "low fat":"Low Fat"
})
```

```python
train.isnull().sum() / len(train) * 100
```

```python
# creating new useful features 
train["Outlet_Age"] = 2025 - (train["Outlet_Establishment_Year"])
```

```python
#  applying low variance filter

numeric = train[["Item_Weight", 'Item_Visibility', 'Item_MRP', 'Outlet_Establishment_Year']]

var = numeric.var(numeric_only=True)
variable = []

for i in range(0, len(var)):
    if var[i] >= 10:
        variable.append(numeric[i])
print(variable)
```
